# 2. Ingenieria de Caracteristicas (feature engineering)

```python
import pandas as pd 
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# cargar los datos
df = pd.read_csv('data/titanic_clean.csv')

x = df[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']]

# Definir columnas 
numerical_cols = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
categorical_cols = ['Sex', 'Embarked']

# Crear el tranformador de columnas 
processor = ColumnTransformer(
   transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
   ]
)

# Ajustar y transformar una muestra para ver el resultado
x_transformed = processor.fit_transform(x)
print("Datos originales")
print(x.iloc[0])
print("Datos transformados", x_transformed[0])
```
